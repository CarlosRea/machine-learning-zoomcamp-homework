```python
# 1. Mediana inicial
median_before = df['horsepower'].median()
print(f"Mediana inicial: {median_before}")
```

```python
# 2. Valor más frecuente (moda)
mode_hp = df['horsepower'].mode()[0]
print(f"Valor más frecuente (moda): {mode_hp}")
```

```python
# 3. Imputar los valores nulos con la moda
df_filled = df.copy()
df_filled['horsepower'] = df_filled['horsepower'].fillna(mode_hp)
```

```python
# 4. Mediana tras imputar
median_after = df_filled['horsepower'].median()
print(f"Mediana tras fillna: {median_after}")
```

```python
# 5. Comprobación
if median_after > median_before:
    print("Respuesta: Yes, it increased")
elif median_after < median_before:
    print("Respuesta: Yes, it decreased")
else:
    print("Respuesta: No")
```

```python
# 1. Seleccionar los coches de Asia
df_asia = df[df['origin'] == 'Asia']
# 2. Seleccionar columnas vehicle_weight y model_year, tomar las primeras 7 filas y extraer el array X
X = df_asia[['vehicle_weight', 'model_year']].head(7).values

# 3. Multiplicación matricial: X^T * X
XTX = X.T.dot(X)

# 4. Invertir XTX
XTX_inv = np.linalg.inv(XTX)

# 5. Crear el array objetivo y
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

# 6. Calcular w = (X^T * X)^(-1) * X^T * y
w = XTX_inv.dot(X.T).dot(y)

# 7. Suma de todos los elementos del vector w
result_sum = w.sum()
```

```python
print("Matriz X (primeras 7 filas):\n", X)
print("\nVector de pesos w:", w)
print(f"\nSuma de los elementos de w: {result_sum}")
print(f"Redondeado: {round(result_sum, 3)}")
```
